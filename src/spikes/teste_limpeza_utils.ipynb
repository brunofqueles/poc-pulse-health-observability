# Spike — validação das funções de limpeza

Confirma que `limpeza_utils.py` reverte corretamente cada tipo de sujeira gerado por `sujeira_intencional.py`, incluindo as variações de nulo.

Ver ADR-002 (streaming vs. batch — Silver usa MERGE, não streaming).

```python
from datetime import date
from src.transformacao.limpeza_utils import (
    parse_data_suja,
    parse_numero_sujo,
    limpar_texto,
    parse_nulo_variado,
)

# Datas — os dois formatos que a sujeira gera
print("Data formato 1:", parse_data_suja("29/07/2026"))
print("Data formato 2:", parse_data_suja("2026-07-29"))
print("Data nula (N/A):", parse_data_suja("N/A"))
print("Data nula (None):", parse_data_suja(None))

# Números — padrão brasileiro
print("Número:", parse_numero_sujo("1.250,50"))
print("Número nulo:", parse_numero_sujo("NULL"))

# Texto — caixa/espaço variando
print("Texto 1:", repr(limpar_texto("APROVADO")))
print("Texto 2:", repr(limpar_texto(" Aprovado ")))
print("Texto 3:", repr(limpar_texto("aprovado")))

# Nulo variado genérico
print("Nulo genérico (vazio):", repr(parse_nulo_variado("")))
print("Nulo genérico (valor real):", repr(parse_nulo_variado("CLI-0001")))
```
