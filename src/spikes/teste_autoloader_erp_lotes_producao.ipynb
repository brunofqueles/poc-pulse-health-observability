# Spike — Autoloader isolado (erp_lotes_producao)

Testa três incertezas antes de generalizar a ingestão para as 11 tabelas:
- `cloudFiles` funciona sem restrição no compute serverless da Free Edition?
- `pathGlobFilter` isola corretamente uma tabela específica dentro da pasta
  do sistema (que tem várias tabelas na mesma partição `data=`)?
- `inferColumnTypes=false` realmente resulta em Bronze 100% string (ADR-001)?

Referências: ADR-001 (Landing Zone), ADR-002 (streaming só Landing→Bronze).

```python
caminho_landing_erp = "/Volumes/poc_pulse_observability/landing/raw/erp"
caminho_schema_autoloader = "/Volumes/poc_pulse_observability/landing/raw/_autoloader_schema/erp_lotes_producao"

df_stream = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.inferColumnTypes", "false")
    .option("cloudFiles.schemaLocation", caminho_schema_autoloader)
    .option("pathGlobFilter", "erp_lotes_producao.json")
    .load(caminho_landing_erp)
)

df_stream.printSchema()
```

```python
caminho_checkpoint = "/Volumes/poc_pulse_observability/landing/raw/_autoloader_checkpoint/erp_lotes_producao"

query = (
    df_stream.writeStream
    .format("delta")
    .option("checkpointLocation", caminho_checkpoint)
    .trigger(availableNow=True)
    .toTable("poc_pulse_observability.bronze.erp_lotes_producao")
)

query.awaitTermination()
print("Status final:", query.lastProgress)
```
