# Visão Geral — Tabelas Silver

Inspeção visual das 11 tabelas Silver: colunas, tipos e volume de linhas, para conferência antes de avançar para a Gold.

Não faz parte do pipeline — é uma verificação pontual, não roda como parte do Job diário.

```python
df_colunas = spark.sql("""
    SELECT table_name, column_name, data_type, ordinal_position
    FROM poc_pulse_observability.information_schema.columns
    WHERE table_schema = 'silver'
    ORDER BY table_name, ordinal_position
""")

df_colunas.show(200, truncate=False)
```

```python
tabelas_silver = [
    "erp_lotes_producao", "erp_posicoes_estoque", "erp_notas_expedicao",
    "crm_pedidos", "crm_itens_pedido", "crm_atendimento",
    "tms_remessas", "tms_leituras_temperatura", "tms_comprovantes_entrega",
    "financeiro_faturas", "financeiro_contas_receber",
]

for tabela in tabelas_silver:
    total = spark.table(f"poc_pulse_observability.silver.{tabela}").count()
    print(f"{tabela}: {total} linhas")
```

```python
for tabela in ["financeiro_faturas", "financeiro_contas_receber"]:
    total = spark.table(f"poc_pulse_observability.silver.{tabela}").count()
    chave = "fatura_id" if tabela == "financeiro_faturas" else "conta_receber_id"
    distintos = spark.table(f"poc_pulse_observability.silver.{tabela}").select(chave).distinct().count()
    print(f"{tabela}: {total} linhas, {distintos} valores distintos de {chave}")
```
